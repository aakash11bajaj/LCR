#### **Question 2:** 
#### Data-visualization

Let's create plots to visualize the relationships between the features (e.g., alcohol, malic acid, ash...).


```python
def plot_feature_pairs(data, feature_names, color_labels=None, title_prefix=''):
    """
    Helper function to create scatter plots for all possible pairs of features.
    
    Parameters:
    - data: DataFrame containing the features to be plotted.
    - feature_names: List of feature names to be used in plotting.
    - color_labels: Optional. Cluster or class labels to color the scatter plots.
    - title_prefix: Optional. Prefix for plot titles to distinguish between different sets of plots.
    """
    # Create a figure for the scatter plots
    plt.figure(figsize=(60, 60))
    
    # Counter for subplot index
    plot_number = 1
    
    # Loop through each pair of features
    for i in range(len(feature_names)):
        for j in range(i + 1, len(feature_names)):
            plt.subplot(len(feature_names)-1, len(feature_names)-1, plot_number)
            
            # Scatter plot colored by labels if provided
            if color_labels is not None:
                plt.scatter(data[feature_names[i]], data[feature_names[j]], 
                            c=color_labels, cmap='viridis', alpha=0.7)
            else:
                plt.scatter(data[feature_names[i]], data[feature_names[j]], alpha=0.7)
            
            plt.xlabel(feature_names[i])
            plt.ylabel(feature_names[j])
            plt.title(f'{title_prefix}{feature_names[i]} vs {feature_names[j]}')
            
            # Increment the plot number
            plot_number += 1

    # Adjust layout to prevent overlap
    plt.tight_layout()

    # Show the plot
    plt.show()

# Get feature names
feature_names = wine_df.columns

# Use the helper function to plot scatter plots without coloring by cluster labels
plot_feature_pairs(wine_df, feature_names, title_prefix='Original Data: ')
```

**Question:**

The plot shows scatterplots for every variable pair available in the dataset- wine_df. Upon inspection of each pair of relationship, we observe that some of the bivariate scatterplots show either positive/negative tending correlations, or little/no correlation between the pair of variables. 
The various variables can be clustered together to be able to distinguish between the different types of wine. 

#### **Question 3:** 
#### Data cleaning

```python
# Initialize the StandardScaler
scaler = StandardScaler()

# Scale all the features in the dataset
scaled_features = scaler.fit_transform(wine_df)

# Create a new DataFrame with scaled features
scaled_wine_df = pd.DataFrame(scaled_features, columns=wine_data.feature_names)

# Display the first few rows of the scaled DataFrame
print(scaled_wine_df.head())
```

Why is it important to standardize the features of a dataset before applying clustering algorithms like K-Means? Discuss the implications of using unstandardized data in your analysis.

The standardization process converts every variable within the dataset into a unitless scalar quantity which has been normalized (centred around the variable mean and SD). This process allows for the various datapoints to be comparable and usable for next steps such as K-means clustering. It also inhibits the domination of one variable in the clustering analysis.  Not implementing any standardization technique prior to the analysis can lead to inaccurate estimation of results, failure of model convergence and/ or misinterpretation of the final results. Standardization is  practice that ensures reliability and reproducibility of results across different runs. 

#### **Question 4:** 
#### K-means clustering 

Apply the K-Means clustering algorithm to the Wine dataset. Choose the value 3 for the number of clusters (`k=3`) and fit the model. Assign cluster labels to the original data and add them as a new column in the DataFrame.

```python
# Perform K-means clustering (k=3)
kmeans = KMeans(n_clusters=3, random_state=11)
clusters = kmeans.fit(scaled_wine_df)

# Create a copy of the DataFrame for clustering and modification
clustered_wine_data = scaled_wine_df.copy()

# Add cluster labels to the copied DataFrame
clustered_wine_data['Cluster'] = clusters.labels_

# Display the updated DataFrame with the 'Cluster' column
clustered_wine_data
```

```python
# Use the helper function to plot scatter plots, colored by cluster labels
plot_feature_pairs(clustered_wine_data, feature_names, color_labels=clustered_wine_data['Cluster'], title_prefix='Clustered Data: ')
```

We chose `k=3` for the number of clusters arbitrarily. However, in a real-world scenario, it is important to determine the optimal number of clusters using appropriate methods.

#### **Question 5:** 
#### Bootstrapping 

Implement bootstrapping on the mean of `color intensity`. Generate 10000 bootstrap samples, calculate the mean for each sample, and compute a 90% confidence interval.

```python
wine_df['color_intensity'].mean()
```

```python
np.random.seed(123)
bootstrap_samples = [] # empty list

for i in range(10_000):
    sample = wine_df.sample(n=178, replace=True)
    sample = sample.assign(replicate=i)
    bootstrap_samples.append(sample)
```

```python
boot10000 = pd.concat(bootstrap_samples)
```

```python
mean_color_intensity = boot10000.groupby("replicate")['color_intensity'].mean().reset_index(name= "sample_mean")
mean_color_intensity
```

```python
# bootstap mean 
mean_color_intensity_boot = mean_color_intensity['sample_mean'].mean()
mean_color_intensity_boot
```

```python

#confidence interval
lower_bound = mean_color_intensity['sample_mean'].quantile(0.1)
upper_bound = mean_color_intensity['sample_mean'].quantile(0.9)
```

```python
# Display the result
print(f"Mean of Color Intensity (bootstrap): {round(mean_color_intensity_boot,3)}")
print(f"90% Confidence Interval of Mean Color Intensity (bootstrap):", round(lower_bound, 4), ",", round(upper_bound, 4),")")
```

**Question:**
- Why do we use bootstrapping in this context? What does it help us understand about the mean?

Bootstrapping is used to calculate an estimate of population parameters (eg mean of colour_intensity) and its variability in the sampling distribution. This process helps us understand the underlying sampling distribution for the variable. In cases where we have limited sample, bootstrapping can help us with the reliability of the results by estimating the properties of the sampling distribution.

**Question:**
- What is the purpose of calculating the confidence interval from the bootstrap samples? How does it help us interpret the variability of the estimate?

The Confidence interval from the bootstrap sample shows the variability in the sampling distribution for the parameter of interest. So, a 90% Bootstrap CI will tell us that that there is 90% chance that the true population parameter value lies within the interval.

The sample mean for Color_intensity is 5.058, while for bootstrap (10,000 samples) mean (90% CI) is 5.059 (4.8377 , 5.281). The sample mean estimate and the bootstrap mean estimate very close to one another implying that there is stability and reliability in the estimates.
Additionally, the 90% CI also is quite narrow implying low variability in sampling distribution for the mean for color_intensity.
